## 问题 1

根据下列递归公式, 计算下列 $T(n)$ 对应的的渐近上界。要求所求的边界尽可能的紧（tight）, 请写明步骤。

1. $T(1)=1; T(n)=T(n/4)+1$ for $n>1$
2. $T(1)=1;T(n)=3T(n/3)+n^2$ for $n>1$
3. $T(1)=1;T(n)=T(2n/3)+1$ for $n>1$
4. $T(1)=1;T(n)=5T(n/4)+n$ for $n>1$
5. $T(n)=1 \ for\ n \le 2 ; T(n)=T(\sqrt{n})+1 \ for \ n>2$

---
**解答：**

对于渐近上界的问题，主要用到的方法是主定理，递归树法，变量替换法。

**题目 1: $T(1)=1$；$T(n)=T(n/4)+1$（$n>1$）**
1. **识别递归类型**：符合主定理标准形式 $T(n) = aT(n/b) + f(n)$，其中 $a = 1, b = 4, f(n) = 1$。
2. **应用主定理**：计算 $n^{\log_b a} = n^{\log_4 1} = n^0 = 1$。
3. **判断情况**：$f(n) = \Theta(1) = \Theta(n^{\log_b a})$，属于情况 2。
4. **结论**：$T(n) = \Theta(n^{\log_b a} \log n) = \Theta(1 \cdot \log n) = \boldsymbol{\Theta(\log n)}$。

**题目 2: $T(1)=1$；$T(n)=3T(n/3)+n^2$（$n>1$）**
1. **识别参数**：$a = 3, b = 3, f(n) = n^2$。
2. **计算比较函数**：$n^{\log_b a} = n^{\log_3 3} = n^1 = n$。
3. **判断情况**：$f(n) = n^2 = \Omega(n^{1+\epsilon})$（取 $\epsilon = 1$），属于情况 3。
4. **验证正则条件**：$a f(n/b) = 3(n/3)^2 = n^2/3 \le (1/3)n^2$，满足 $c < 1$ 的条件。
5. **结论**：$T(n) = \Theta(f(n)) = \boldsymbol{\Theta(n^2)}$。

**题目 3: $T(1)=1$；$T(n)=T(2n/3)+1$（$n>1$）**
1. **识别参数**：$a = 1, b = 3/2, f(n) = 1$。
2. **计算比较函数**：$n^{\log_b a} = n^{\log_{3/2} 1} = n^0 = 1$。
3. **判断情况**：$f(n) = \Theta(1) = \Theta(n^{\log_b a})$，属于情况 2。
4. **结论**：$T(n) = \Theta(1 \cdot \log n) = \boldsymbol{\Theta(\log n)}$。

**题目 4: $T(1)=1$；$T(n)=5T(n/4)+n$（$n>1$）**
1. **识别参数**：$a = 5, b = 4, f(n) = n$。
2. **计算比较函数**：$n^{\log_b a} = n^{\log_4 5} \approx n^{1.16}$。
3. **判断情况**：$f(n) = n = O(n^{\log_4 5 - \epsilon})$，属于情况 1。
4. **结论**：$T(n) = \Theta(n^{\log_4 5}) = \boldsymbol{\Theta(n^{\log_4 5})}$。

**题目 5: $T(n)=1$（$n \le 2$）；$T(n)=T(\sqrt{n})+1$（$n>2$）**
1. **变量替换**：令 $m = \log_2 n$，则 $n = 2^m, \sqrt{n} = 2^{m/2}$。
2. **新递归式**：设 $S(m) = T(2^m)$，则 $S(m) = S(m/2) + 1$。
3. **应用主定理**：对 $S(m)$ 使用主定理，$a=1, b=2, f(m)=1$，得到 $S(m) = \Theta(\log m)$。
4. **代回变量**：$T(n) = S(\log_2 n) = \Theta(\log (\log_2 n)) = \boldsymbol{\Theta(\log \log n)}$。

## 问题 2

观察下方的分治算法（divide-and-conquer algorithm）的伪代码, 回答下面问题

```latex
DoSomething(A,p,r)
-----
n := r-p+1
if n=2 and A[p]>A[r] then
    swap A[p] and A[r]
else if n >= 3 then
    m = ceil(2n/3)
    DoSomething(A,p,p+m-1)
    DoSomething(A,r-m+1,r)
    DoSomething(A,p,p+m-1)  
    
---
first call: DoSomething(A,1,n)
```

note：$ceil(2n/3)=\left\lceil {2n/3} \right\rceil$；$:=$ 表示赋值, 等价于 $\to$；A是一个包含n的整数元素的数组, 

1. 写出该算法时间复杂度的递归公式, 并求解其对应的渐进表示
2. 描述一下该算法的功能, 并判断是否是最高效的解决方案
3. \* 使用python编程实现上述算法或其对应的更高效版本的算法

**1. 时间复杂度的递归公式和渐近表示**

**(1) 递归公式**

观察代码可以看出，对于规模为 $n$ 的数组：
* 当 $n = 2$ 时，执行一次比较和可能的交换，时间开销为常数，即 $O(1)$。
* 当 $n \ge 3$ 时，算法将其分为规模大小约为 $2n/3$ 的子问题，并且一共进行了 3 次递归调用（前三分之二，后三分之二，再前三分之二）。
* 非分割和计算 $m$ 的时间开销同样为常数 $O(1)$。

因此，时间复杂度的递归公式为：
$$T(n) = 3T(2n/3) + O(1)$$
边界条件：$T(2)=O(1), T(1)=O(1)$。

**(2) 渐近表示（主定理求解）**

将递归公式对应到主定理标准形式 $T(n) = aT(n/b) + f(n)$：
* $a = 3$（3次递归调用）
* $b = 3/2 = 1.5$（子问题规模缩小比例）
* $f(n) = O(1) = n^0$（额外工作量）

计算基准值：$n^{\log_b a} = n^{\log_{1.5} 3} \approx n^{2.7095}$。
比较 $f(n) = n^0$ 与 $n^{2.7095}$，符合主定理情况 1（叶子节点主导）。

结论：$$T(n) = \Theta(n^{\log_{1.5} 3}) \approx \boldsymbol{\Theta(n^{2.71})}$$

**2. 算法功能与效率评估**

* **功能描述**：该算法是 **Stooge Sort（臭皮匠排序）**。它通过重叠分治（递归处理前 2/3 和后 2/3）来确保数组有序。
* **效率评估**：**不是**最高效的。其复杂度 $\Theta(n^{2.71})$ 远差于冒泡排序的 $O(n^2)$，更无法与归并排序或快速排序的 $O(n \log n)$ 相比。

````python
def merge_sort(A):
    """
    更高效的替代版本：归并排序
    """
    if len(A) <= 1:
        return A
        
    mid = len(A) // 2
    left_half = merge_sort(A[:mid])
    right_half = merge_sort(A[mid:])
    
    return merge(left_half, right_half)

def merge(left, right):
    sorted_array = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            sorted_array.append(left[i])
            i += 1
        else:
            sorted_array.append(right[j])
            j += 1
    sorted_array.extend(left[i:])
    sorted_array.extend(right[j:])
    return sorted_array

# 测试
arr = [5, 2, 9, 1, 7, 6, 3]
print("排序结果:", merge_sort(arr))

# your algorithm time complexity is: O(n log n)
````

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

**1. 设计算法思路**

**理解与转化：**
从示例可以看出，最大利润对应的是收益序列中某一段连续子数组的和。因此，问题转化为在数组中找到**连续子数组的最大和**，并记录对应的起始（买入）和结束（卖出）索引。

* **目标**：找到 $(i, j)$ 使得 $\sum_{k=i}^{j} profits[k-1]$ 最大。
* **核心逻辑**：需寻找长度至少为 1 的连续子数组（题目要求“不同日子”，即至少跨越一个收益周期）。

**方案一：暴力法 (O(n²))**
使用双重循环遍历所有可能的买入日 $i$ 和卖出日 $j$，计算区间收益并更新最大值。这是基准算法。

**方案二：更高效的 Kadane 算法变种 (O(n))**
* **核心思想**：动态规划。维护一个“以当前元素结尾的局部最优解” `current_sum`。 
* **决策逻辑**：对于元素 $profits[i]$，比较 `current_sum + profits[i]` 与 `profits[i]`。若前者更大，则扩展子数组；若后者更大，则说明之前的累积已成为负担，应从当前天重新开始（重置 `temp_buy`）。
* **状态追踪**：始终用局部最优更新全局最优 `max_sum`，并记录对应的 `buy_day` 和 `sell_day`。

````python
def max_profit_bruteforce(profits):
    """暴力法实现 O(n^2)"""
    n = len(profits)
    max_p = float('-inf')
    buy, sell = -1, -1
    for i in range(n):
        cur_sum = 0
        for j in range(i, n):
            cur_sum += profits[j]
            if cur_sum > max_p:
                max_p, buy, sell = cur_sum, i + 1, j + 1
    return max_p, buy, sell

def max_profit_kadane(profits):
    """Kadane 算法优化 O(n)"""
    n = len(profits)
    if n == 0: return 0, -1, -1
    max_sum = cur_sum = profits[0]
    buy = sell = temp_buy = 1
    
    for i in range(1, n):
        if cur_sum + profits[i] > profits[i]:
            cur_sum += profits[i]
        else:
            cur_sum = profits[i]
            temp_buy = i + 1
            
        if cur_sum > max_sum:
            max_sum, buy, sell = cur_sum, temp_buy, i + 1
    return max_sum, buy, sell

# 示例测试
profits_data = [3, 2, 1, -7, 5, 2, -1, 3, -1]
res, b_day, s_day = max_profit_kadane(profits_data)
print(f"最大利润: {res}, 买入: 第{b_day}天, 卖出: 第{s_day}天")

# your algorithm time complexity is: O(n)
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

**1. 算法设计思路（找一个局部最小：二分查找法）**

利用地形类比：数组中的数值代表海拔。由于 $A[0] > A[1]$（出发下坡）且 $A[n-2] < A[n-1]$（终点前上坡），中间必然存在至少一个“山谷”（局部最小）。

* **核心逻辑**：通过二分查找缩小范围。取 $mid$，若 $A[mid]$ 比左右邻居都小，则命中；否则，若左邻居更小，说明左侧存在下降趋势，向左半区搜索；若右邻居更小，则向右半区搜索。
* **时间复杂度**：$O(\log n)$，因为每次排除了一半的搜索空间。

**2. 算法设计思路（找所有局部最小：线性扫描法）**

由于需要找出“所有”局部最小，二分查找的剪枝策略不再适用（一个区间内可能隐藏多个山谷）。

* **核心逻辑**：遍历数组从索引 $1$ 到 $n-2$，对每个元素检查其是否同时小于左右邻居。如果是，则将其计入结果列表。
* **时间复杂度**：$O(n)$，因为必须检查每一个潜在的候选点以防遗漏。

````python
def find_one_local_minimum(arr):
    """二分查找 O(log n) 找其中一个"""
    n = len(arr)
    left, right = 1, n - 2
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] < arr[mid - 1] and arr[mid] < arr[mid + 1]:
            return arr[mid]
        elif arr[mid - 1] < arr[mid]:
            right = mid - 1
        else:
            left = mid + 1
    return None

def find_all_local_minima(arr):
    """线性扫描 O(n) 找所有"""
    n = len(arr)
    result = []
    for i in range(1, n - 1):
        if arr[i] < arr[i - 1] and arr[i] < arr[i + 1]:
            result.append(arr[i])
    return result

# 测试
A = [9, 3, 7, 2, 1, 4, 5]
print(f"找一个局部最小: {find_one_local_minimum(A)}")
print(f"找所有局部最小: {find_all_local_minima(A)}")

# your algorithm time complexity is: 1. O(log n), 2. O(n)
````

## 问题 5

给定包含n个不同数字的一组数, 寻找一种基于比较的算法在这组数中找到k个最小的数字, 并按顺序输出它们。

1. 将n个数先进行排序, 然后按顺序输出最小的k个数。要求：选择合适的排序算法实现上述操作, 计算算法时间复杂度, 并使用python编程实现。
2. 建立一个包含这n个数的堆（heap）, 并且调用 k 次Extract-min 按顺序输出最小的k个数。使用往空堆中不断插入元素的方法建立堆, 分析这种方法建堆的时间复杂度, 并使用python编程实现
3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

### 第 1 题：全局排序法
1. **算法设计思路**：直接将所有数据从小到大排序（使用 Python 内置的 Timsort，$O(n \log n)$），然后截取前 $k$ 个元素。
2. **时间复杂度**：排序 $O(n \log n)$，输出 $O(k)$，总计 **$O(n \log n)$**。

### 第 2 题：建立最小堆法
1. **算法设计思路**：模拟从空堆开始，通过 $n$ 次插入操作建立最小堆。随后连续调用 $k$ 次 `Extract-min`（弹出堆顶）获取前 $k$ 小元素。
2. **时间复杂度分析**：
* **建堆（插入法）**：插入第 $i$ 个元素需 $O(\log i)$，总复杂度 $\sum \log i = \log(n!) \approx \boldsymbol{O(n \log n)}$。
* **提取过程**：$k$ 次弹出，每次调整堆耗时 $O(\log n)$，总计 $O(k \log n)$。
* **总计**：**$O(n \log n)$**。

### 第 3 题：海量数据处理（内存受限）
1. **算法设计思路**：维护一个容量为 $k$ 的**最大堆**。遍历海量数据，若新数小于堆顶（“守门员”），则替换堆顶并重整堆。此方法只需 $O(k)$ 空间。
2. **复杂度分析**：时间复杂度为 **$O(n \log k)$**，空间复杂度为 **$O(k)$**。在 $k \ll n$ 时效率极高且符合内存限制。

````python
import heapq

def find_top_k_by_sort(arr, k):
    """1. 全局排序法 O(n log n)"""
    return sorted(arr)[:k]

def find_top_k_by_min_heap(arr, k):
    """2. 最小堆法（模拟逐个插入建堆） O(n log n)"""
    min_heap = []
    for num in arr:
        heapq.heappush(min_heap, num)
    return [heapq.heappop(min_heap) for _ in range(k)]

def find_top_k_massive_data(arr, k):
    """3. 海量数据法（最大堆过滤器） O(n log k)"""
    max_heap = []
    for num in arr:
        if len(max_heap) < k:
            heapq.heappush(max_heap, -num)
        else:
            if num < -max_heap[0]:
                heapq.heapreplace(max_heap, -num)
    res = [-x for x in max_heap]
    return sorted(res)

# 测试
arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3
print(f"排序法结果: {find_top_k_by_sort(arr, k)}")
print(f"最小堆结果: {find_top_k_by_min_heap(arr, k)}")
print(f"海量数据法: {find_top_k_massive_data(arr, k)}")

# your algorithm time complexity is: 1. O(n log n), 2. O(n log n), 3. O(n log k)
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

**算法设计思路：双指针模拟栈（原地操作）**

我们将原数组的前半部分视为一个“逻辑栈”，通过两个指针高效模拟碰撞过程：

* **核心逻辑**：指针 `top` 标记已整理好的“存活区”末端，指针 `i` 遍历原数组。
* **碰撞条件**：只有当 `top >= 0` 且 `A[top] > 0`（向右运动）且当前星球 `p < 0`（向左运动）时，才会发生碰撞。
* **规则处理**：
  1. 若 $|A[top]| < |p|$，栈顶星球碎裂，`top` 下移并继续与新栈顶比对。
  2. 若 $|A[top]| \ge |p|$，根据规则（质量相等时向左的消失），当前星球 `p` 碎裂，停止比对。
* **更新存活区**：若当前星球经历碰撞后未碎裂（或未发生碰撞），则将其移至 `A[++top]` 位置。

**复杂度分析：**
* **时间复杂度**：$O(n)$。每个星球最多入栈和出栈各一次。
* **空间复杂度**：$O(1)$。直接在原数组上进行读写操作（原地修改）。

````python
def planet_collision_optimized(A):
    # top 指向当前存活星球的最右侧索引
    top = -1 
    
    for p in A:
        is_destroyed = False
        # 碰撞逻辑：栈顶向右(>0) 且 当前向左(<0)
        while top >= 0 and A[top] > 0 and p < 0:
            if abs(A[top]) < abs(p):
                top -= 1 # 栈顶碎裂，继续向前比对
                continue
            else:
                is_destroyed = True # 栈顶更大或质量相等，当前星球碎裂
                break
        
        if not is_destroyed:
            top += 1
            A[top] = p
            
    return A[:top + 1]

# 验证
A_test = [23, -8, 9, -3, -7, 9, -23, 22]
print(f"星球存续情况: {planet_collision_optimized(A_test)}")

# your algorithm time complexity is: O(n)
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

**解决 0-1 背包问题的最优解法是动态规划。**

我们将大问题拆解为规模更小的子问题来逐步求解：

1. **状态定义**：
   定义 $dp[i][w]$ 表示前 $i$ 块大理石在小推车承重限制为 $w$ 时，所能装载的最大总价值。

2. **状态转移方程**：
   对于当前大理石（重量 $weight$，价值 $value$），我们有两种选择：
   * **不装**：价值维持 $dp[i-1][w]$。
   * **装入**：前提是 $w \ge weight$，价值为 $dp[i-1][w-weight] + value$。
   * **方程**：$$dp[i][w] = \max(dp[i-1][w], dp[i-1][w-weight] + value)$$

3. **方案溯源**：
   为了找出具体选择了哪些大理石，我们使用二维 DP 表。如果 $dp[i][w] \ne dp[i-1][w]$，说明第 $i$ 块大理石被选中了。

**复杂度分析：**
* **时间复杂度**：$O(N \cdot W)$。其中 $N$ 是大理石数量（4），$W$ 是推车最大承重（9）。
* **空间复杂度**：$O(N \cdot W)$。使用二维数组记录所有中间状态以支持方案回溯。

````python
def knapsack_01_with_items(items, capacity):
    """
    计算 0-1 背包最大价值并返回选中的物品编号
    """
    n = len(items)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    # 1. 动态规划填表
    for i in range(1, n + 1):
        weight, value = items[i - 1]
        for w in range(1, capacity + 1):
            if weight <= w:
                dp[i][w] = max(dp[i - 1][w], dp[i - 1][w - weight] + value)
            else:
                dp[i][w] = dp[i - 1][w]

    max_value = dp[n][capacity]

    # 2. 逆向回溯，找出具体物品
    chosen_items = []
    w = capacity
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i - 1][w]:
            chosen_items.append(i)
            w -= items[i - 1][0]

    chosen_items.reverse() 
    return max_value, chosen_items

# 测试数据
marbles = [(5, 10), (4, 40), (6, 30), (3, 50)]
max_weight = 9
val, picked = knapsack_01_with_items(marbles, max_weight)

print(f"最大价值: {val}")
print(f"选择的大理石编号: {picked}")

# your algorithm time complexity is: O(N*W)
````

给你二叉搜索树的根节点 root ，该树中的恰好两个节点的值被错误地交换。请在不改变其结构的情况下，恢复这棵树 。设计算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]，

输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

**算法设计思路：BST 中序遍历 + Morris 遍历优化**

**1. 核心性质**：
二叉搜索树（BST）的中序遍历序列应是**严格递增**的。若有两个节点被错误交换，遍历序列中会出现“降序对”：
* **相邻交换**：产生一个降序对，错误的两个节点即为该对的两个元素。
* **非相邻交换**：产生两个降序对，错误节点为第一个降序对的前者和第二个降序对的后者。

**2. Morris 遍历实现 $O(1)$ 空间优化**：
为了避免 $O(H)$ 的递归栈开销，使用 Morris 遍历利用叶子节点的右空指针建立临时“回程桥”：
* **搭桥**：找到当前节点左子树的最右节点（前驱），将其 `right` 指向当前节点。
* **回溯与拆桥**：通过桥回到根节点后，处理节点逻辑并恢复（置空）`right` 指针。

**3. 复杂度分析**：
* **时间复杂度**：$O(N)$。每条边最多被访问两次（搭桥和拆桥）。
* **空间复杂度**：$O(1)$。仅使用了常数个指针变量（`first`, `second`, `prev`, `curr`, `pred`）。

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def recoverTree(root):
    """
    使用 Morris 遍历原地恢复 BST
    """
    first = second = prev = None
    curr = root
    
    while curr:
        if not curr.left:
            # 访问节点逻辑
            if prev and prev.val > curr.val:
                if not first: first = prev
                second = curr
            prev = curr
            curr = curr.right
        else:
            pred = curr.left
            while pred.right and pred.right != curr:
                pred = pred.right
            
            if not pred.right:
                pred.right = curr # 搭桥
                curr = curr.left
            else:
                pred.right = None # 拆桥
                # 访问节点逻辑
                if prev and prev.val > curr.val:
                    if not first: first = prev
                    second = curr
                prev = curr
                curr = curr.right

    if first and second:
        first.val, second.val = second.val, first.val

# your algorithm time complexity is: O(N)
# your algorithm space complexity is: O(1)
````

## 问题 9 \*
给定 `n` 个非负整数表示每个宽度为 `1` 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。



<div style="text-align:center;width:800px;"> <img alt="" src="./fig/hw3q6.png"> </div>

请分析算法设计思路，计算时间复杂度，并基于python编程实现

**算法设计思路：双指针法（基于木桶效应）**

**1. 核心原理**：
每一根柱子 $i$ 上方能接的水量取决于其左侧最高柱子和右侧最高柱子中的较小值（木桶短板），公式为：
$$\text{water}[i] = \max(0, \min(\text{left\_max}[i], \text{right\_max}[i]) - \text{height}[i])$$

**2. 双指针优化思路**：
为了将空间复杂度从 $O(n)$ 降至 $O(1)$，我们让 `left` 和 `right` 指针从两端向中间移动：
* 如果 `height[left] < height[right]`，由于右侧已经有一个更高的“墙”保证了水位，当前 `left` 位置的蓄水量仅取决于左侧已知的 `left_max`。我们将计算结果累加并右移 `left`。
* 反之，如果 `height[right] \le height[left]`，则处理 `right` 位置，其蓄水量取决于右侧的 `right_max`。

**3. 复杂度分析**：
* **时间复杂度**：$O(n)$。数组仅被线性遍历一次。
* **空间复杂度**：$O(1)$。只使用了常数个辅助变量。

````python
def trap_rain_water(height):
    if not height:
        return 0
    
    left, right = 0, len(height) - 1
    left_max = right_max = 0
    total_water = 0
    
    while left < right:
        if height[left] < height[right]:
            # 处理左侧：受限于 left_max
            if height[left] >= left_max:
                left_max = height[left]
            else:
                total_water += left_max - height[left]
            left += 1
        else:
            # 处理右侧：受限于 right_max
            if height[right] >= right_max:
                right_max = height[right]
            else:
                total_water += right_max - height[right]
            right -= 1
            
    return total_water

# 测试用例
h = [0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
print(f"高度图: {h}")
print(f"总接雨水量: {trap_rain_water(h)}")

# your algorithm time complexity is: O(n)
# your algorithm space complexity is: O(1)
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

> 例如，从根节点到叶节点的路径 1 -> 2 -> 3 表示数字 123 。

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:
---
1.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-1.jpg"> </div>

 输入：root = [1,2,3], 输出：25

解释：从根到叶子节点路径 1->2 代表数字 12，1->3 代表数字 13，因此和为 12 + 13 = 25 。

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

**算法设计思路：深度优先搜索（DFS）**

**1. 核心逻辑（数字拼接原理）**：
从根节点向叶子节点遍历时，每向下走一层，相当于将当前的路径数字扩大 10 倍并加上新节点的值。例如路径 $1 \to 2 \to 3$ 的生成过程为：$(1 \times 10 + 2) \times 10 + 3 = 123$。

**2. DFS 执行步骤**：
* **递归函数**：定义 `dfs(node, current_sum)`，其中 `current_sum` 记录从根到父节点的路径和。
* **更新状态**：计算当前路径值：`current_sum = current_sum * 10 + node.val`。
* **终止条件**：若当前节点为叶子节点（左右孩子皆为空），则返回当前 `current_sum`。
* **递归分支**：若非叶子节点，则递归计算左、右子树的和并相加。

**3. 复杂度分析**：
* **时间复杂度**：$O(N)$。每个节点仅被访问一次。
* **空间复杂度**：$O(H)$。取决于递归栈深度，平均情况下为 $O(\log N)$，最坏情况（链状树）为 $O(N)$。

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def sumNumbers(root):
    def dfs(node, current_sum):
        if not node: return 0
        
        # 计算当前路径数字
        current_sum = current_sum * 10 + node.val
        
        # 到达叶子节点，返回该路径形成的数字
        if not node.left and not node.right:
            return current_sum
        
        # 返回左右子树路径之和
        return dfs(node.left, current_sum) + dfs(node.right, current_sum)

    return dfs(root, 0)

# 测试示例 2: [4, 9, 0, 5, 1]
root = TreeNode(4, TreeNode(9, TreeNode(5), TreeNode(1)), TreeNode(0))
print(f"根到叶所有路径生成的数字之和: {sumNumbers(root)}")

# your algorithm time complexity is: O(N)
# your algorithm space complexity is: O(H)
````
